```
x=0
menor=0
soma=0
while x<5:
  med= input("Nome do medicamento: ")
  preco=float(input("Preço do medicamento: "))
  soma+=preco
  if x==0:
    menor=preco
    nome_menor=med
  else:
    if preco<menor:
      menor=preco
      nome_menor=med
  x=x+1
media=soma/x
print("Medicamento mais barato: ",nome_menor)
print("Preço do mais barato: R$",menor)
print("Média dos preços informados: R$",media)
```
