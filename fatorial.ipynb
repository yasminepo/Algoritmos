```python
while True:
    num = int(input("Digite um número inteiro positivo: "))
    if num == 0:
        break
    else:
        def fat(num):
            resp = 1
            for i in range(1, num + 1):
                resp *= i
            return resp
        print(f"O fatorial de {num} = {fat(num)}")
```
