```python
print("Faça o cadastro aqui!")
lista1 = []
lista2 = []
quant=0
while True:
    x=input("Digite o nome de usuário: ")
    y=input("Digite a senha do usuário: ")
    lista1.append (x)
    lista2.append (y)
    quant+=1
    resp = input("Deseja realizar outro cadastro [S|N]?: ")
    if resp.upper() == "N":
        break
print("Faça o login aqui!")
usuario=input("Digite o nome do usuário: ")
senha=input("Digite a senha do usuário: ")
i = 0
while i < quant:
    if usuario == lista1[i] and senha == lista2[i]:
        print("Login realizado com sucesso!")
        break
    i += 1
if i == quant:
    print("Usuário e/ou senha incorreto(s)!")
```
