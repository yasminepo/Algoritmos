```python
import random
lista = []
while True:
    lista.append (input("Nome do comprador da rifa: "))
    resp = input("Deseja continuar [S|N]?: ")
    if resp.upper() == "N":
        break
ganhador = random.choice(lista)
print("Nome do ganhador do prêmio:", ganhador)
```
